```python
import cv2
import numpy as np
from random import randint 
import glob
from  tqdm import tqdm
import os 

def rotate_image(image, degrees):
    # Load the image
    # image = cv2.imread(image_path)
    
    # Get the image dimensions
    (h, w) = image.shape[:2]

    # Calculate the center of the image
    center = (w / 2, h / 2)

    # Perform the rotation
    M = cv2.getRotationMatrix2D(center, degrees, 1.0)
    rotated = cv2.warpAffine(image, M, (w, h))
    return rotated

def shear_image(image, shear_degree):

    # Read the image
    # image = cv2.imread(image_path)

    # Convert degrees to radians for transformation
    radians = np.deg2rad(shear_degree)

    # Calculate the shear matrix
    shear_matrix = np.array([[1, np.tan(radians), 0],
                             [0, 1, 0]])

    # Apply the shear transformation
    sheared_image = cv2.warpAffine(image, shear_matrix, (image.shape[1], image.shape[0]))

    return sheared_image

def change_brightness(img, value):
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    h, s, v = cv2.split(hsv)
    v = cv2.add(v,value)
    v[v > 255] = 255
    v[v < 0] = 0
    final_hsv = cv2.merge((h, s, v))
    img = cv2.cvtColor(final_hsv, cv2.COLOR_HSV2BGR)
    return img
```

```python
import os

def rename_images_to_match(folder1, folder2):
    # Get list of image files in each folder
    images1 = sorted([img for img in os.listdir(folder1) if img.lower().endswith(('.png', '.jpg', '.jpeg'))])
    images2 = sorted([img for img in os.listdir(folder2) if img.lower().endswith(('.png', '.jpg', '.jpeg'))])

    # Check if the number of images in both folders is the same
    if len(images1) != len(images2):
        return "The number of images in both folders is not the same."

    # Renaming images in folder 2 to match the names in folder 1
    for img1, img2 in zip(images1, images2):
        src = os.path.join(folder2, img2)
        dst = os.path.join(folder2, img1)
        os.rename(src, dst)

    return "Images in folder 2 have been renamed to match folder 1."

# Example usage
folder1_path = 'D:\\identification_Cracks\\Dataset\\UnCrack\\Anchor'
folder2_path = 'D:\\identification_Cracks\\Dataset\\UnCrack\\Nagative'
result = rename_images_to_match(folder1_path, folder2_path)
print(result)
```
